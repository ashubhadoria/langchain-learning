## JSON Output

```python
from langchain_openai import ChatOpenAI
from langchain_core.pydantic_v1 import BaseModel

class AnswerWithJustification(BaseModel):
    answer: str
    justification: str

llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)
structured_llm = llm.with_structured_output(AnswerWithJustification)

result = structured_llm.invoke("What weighs more a kilogram of steel or a kilogram of feathers?")
result
```

## Runnable interface

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="gpt-3.5-turbo")

completion = model.invoke("Hi there!")
print("invoke: ", completion)

completions = model.batch(["Hi there!", "Hasta la vista!"])
print("batch: " , completions)

for token in model.stream("Adios friend!"):
    print("stream: ")
    print(token)
```

## Imperative Composition

### Sync Execution


```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import chain

model = ChatOpenAI(model="gpt-3.5-turbo")

template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant"),
        ("human", "Question: {question}"),
    ]
)

@chain
def chatbot(values):
    prompt = template.invoke(values)
    return model.invoke(prompt)

chatbot.invoke({"question": "Which model provides LLMs?"})
```

### Streaming Execution

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import chain

model = ChatOpenAI(model="gpt-3.5-turbo")

template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant"),
        ("human", "Question: {question}"),
    ]
)

@chain
def chatbot(values):
    prompt = template.invoke(values)
    for token in model.stream(prompt):
        yield token

for chunk in chatbot.stream({"question": "Which model provides LLMs?"}):
    print(chunk)
```

### Asyc Execution

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import chain

model = ChatOpenAI(model="gpt-3.5-turbo")

template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant"),
        ("human", "Question: {question}"),
    ]
)

@chain
async def chatbot(values):
    prompt = template.invoke(values)
    return await model.ainvoke(prompt)

await chatbot.ainvoke({"question": "Which model provides LLMs?"})
```

## LangChain Expression Language (LCEL)  Composition

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

model = ChatOpenAI(model="gpt-3.5-turbo")

template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant"),
        ("human", "Question: {question}"),
    ]
)

chatbot = template | model

chatbot.invoke({"question": "Which model provides LLMs?"})
```
